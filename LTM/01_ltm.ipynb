```python
from langgraph.store.memory import InMemoryStore
```

```python
store=InMemoryStore()
```

```python
#creating a namespace for our memories
namespace=('user','url')
```

```python
#Creating Memories

#Adding memories to the store
store.put(namespace,"1",{'data':'User likes spaghetti'})
store.put(namespace,"2",{'data':'User is a software engineer'})
```

```python
#creating another namespace for more memories
namespace2=('user','u2')

store.put(namespace2,"1",{'data':'User likes pizza'})
store.put(namespace2,"2",{'data':'User is a data scientist'})
```

```python
#Retrieving memories

#store.get(namespace,key)
store.get(namespace,"1")
```

```python
#Retrieving memories

items=store.search(namespace)

for item in items:
    print(item.value)
```
